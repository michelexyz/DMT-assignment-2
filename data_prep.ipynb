```python
import pandas as pd
```

```python
train_path = "data/training_set_VU_DM.csv"
chunk = pd.read_csv(train_path, chunksize=1000000)
df = pd.concat(chunk)
```

# Dropping columns:

[Dany] dropped because they do not have enough values: 'srch_query_affinity_score', 'visitor_hist_starrating','visitor_hist_adr_usd'



```python
prepped_df = df.drop(['date_time','orig_destination_distance','comp1_rate','comp1_inv','comp1_rate_percent_diff','comp2_rate','comp2_inv','comp2_rate_percent_diff','comp3_rate','comp3_inv','comp3_rate_percent_diff','comp4_rate','comp4_inv','comp4_rate_percent_diff','comp5_rate','comp5_inv','comp5_rate_percent_diff','comp6_rate','comp6_inv','comp6_rate_percent_diff','comp7_rate','comp7_inv','comp7_rate_percent_diff','comp8_rate','comp8_inv','comp8_rate_percent_diff','srch_query_affinity_score', 'visitor_hist_starrating','visitor_hist_adr_usd'],axis=1)
```

```python
prepped_df['relevance'] = prepped_df['click_bool'] + prepped_df['booking_bool']*4
```

```python
path = f"data/basic_dataset.csv"
prepped_df.to_csv(path, index=False)
```
