# Setting up

```python
# IMPORTS
import pandas as pd
import time
```

```python
# PARAMETERS

train_path = "data/basic_dataset.csv"
# for testing: train_path = "data/small_dataset.csv"
stats_path = f"stats/{train_path.split('/')[1][:-4]}_stats.csv"
save_output = True
```

```python
# LOAD DATA

start_time = time.time()
chunk = pd.read_csv(train_path, chunksize=1000000)
print(f"load chunk: {time.time() - start_time:.2f}s")

start_time = time.time()
df = pd.concat(chunk)
print(f"concat df: {time.time() - start_time:.2f}s")
```

# Describing the dataset

### Describing the entire dataset

```python
# save to csv the stats
stats = []

for column in df.columns:
    stats.append(df[column].describe())

df_stats = pd.concat(stats, axis=1).T

if save_output: df_stats.to_csv(stats_path)
```

```python
df_stats.head()
```

```python
for row_name in df_stats.index:
     df_stats.loc[row_name, "count_missing"] = df[row_name].isna().sum()
     
if save_output: df_stats.to_csv(stats_path)
```

```python
# describe the dataframe with describe method with 2 decimals
df.describe().round(2).T
```

### Describing only the booked rowss

```python
temp_df = df[df.booking_bool == 1]
stats = []
for column in temp_df.columns:
    stats.append(temp_df[column].describe())
df_stats = pd.concat(stats, axis=1).T

stats_path = f"stats/{train_path.split('/')[1][:-4]}_stats_booked_rows.csv"
if save_output: df_stats.to_csv(stats_path)    
```

```python
for row_name in df_stats.index:
     df_stats.loc[row_name, "count_missing"] = temp_df[row_name].isna().sum()

if save_output: df_stats.to_csv(stats_path)
```
