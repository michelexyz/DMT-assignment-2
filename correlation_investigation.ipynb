```python
import pandas as pd
from tabulate import tabulate
from matplotlib import pyplot as plt
import scipy
```

```python
# load data
path = "data/training_set_VU_DM.csv"
df = pd.read_csv(path)
```

```python
def do_corr_matrix(corr_method='pearson'):
    f = plt.figure(figsize=(19, 15))
    corr_df = df.corr(method=corr_method)
    plt.matshow(corr_df, fignum=f.number)
    plt.xticks(range(df.select_dtypes(['number']).shape[1]), df.select_dtypes(['number']).columns, fontsize=14, rotation=45)
    plt.yticks(range(df.select_dtypes(['number']).shape[1]), df.select_dtypes(['number']).columns, fontsize=14)
    cb = plt.colorbar()
    cb.ax.tick_params(labelsize=14)
    plt.title('Correlation Matrix', fontsize=16)
    plt.savefig(f"graphs/correlation/corr_{corr_method}_{path.split('/')[-1].split('.')[0]}.png")
    plt.show()
    return corr_df
```

```python
# spearman
spear_corr = do_corr_matrix("spearman")
```

```python
# pearson
pearson_corr = do_corr_matrix("pearson")
```

```python
# kendall
kendall_corr = do_corr_matrix("kendall")
```
